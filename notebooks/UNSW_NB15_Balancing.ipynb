```python
# ============================================================
# UNSW-NB15 - SINGLE CELL MEMORY-SAFE BALANCING PIPELINE
# ============================================================

import os
import gc
import pandas as pd
from sklearn.model_selection import train_test_split

# ------------------------------------------------------------
# 1. MOUNT GOOGLE DRIVE
# ------------------------------------------------------------
from google.colab import drive
drive.mount('/content/drive')

# ------------------------------------------------------------
# 2. DATASET PATH
# ------------------------------------------------------------
dataset_path = "/content/drive/MyDrive/UNSW-NB15"

files = [
    os.path.join(dataset_path, "UNSW-NB15_1.csv"),
    os.path.join(dataset_path, "UNSW-NB15_2.csv"),
    os.path.join(dataset_path, "UNSW-NB15_3.csv"),
    os.path.join(dataset_path, "UNSW-NB15_4.csv")
]

# ------------------------------------------------------------
# 3. CORRECT 49 COLUMN NAMES
# ------------------------------------------------------------
columns = [
    "srcip", "sport", "dstip", "dsport", "proto", "state",
    "dur", "sbytes", "dbytes", "sttl", "dttl", "sloss", "dloss",
    "service", "Sload", "Dload", "Spkts", "Dpkts", "swin", "dwin",
    "stcpb", "dtcpb", "smeansz", "dmeansz", "trans_depth",
    "res_bdy_len", "Sjit", "Djit", "Stime", "Ltime",
    "Sintpkt", "Dintpkt", "tcprtt", "synack", "ackdat",
    "is_sm_ips_ports", "ct_state_ttl", "ct_flw_http_mthd",
    "is_ftp_login", "ct_ftp_cmd", "ct_srv_src", "ct_srv_dst",
    "ct_dst_ltm", "ct_src_ltm", "ct_src_dport_ltm",
    "ct_dst_sport_ltm", "ct_dst_src_ltm", "attack_cat", "Label"
]

# ------------------------------------------------------------
# 4. LOAD THE 4 RAW CSV FILES
#    IMPORTANT: These files have NO HEADER
# ------------------------------------------------------------
print("Loading UNSW-NB15 files...\n")

dfs = []

for file in files:

    print("Loading:", os.path.basename(file))

    temp = pd.read_csv(
        file,
        header=None,
        names=columns,
        encoding="latin1"
    )

    # Create Normal class from missing attack_cat
    temp["attack_cat"] = temp["attack_cat"].fillna("Normal")

    # Clean class names
    temp["attack_cat"] = (
        temp["attack_cat"]
        .astype(str)
        .str.strip()
        .replace({"Backdoor": "Backdoors"})
    )

    print("Rows:", len(temp))

    dfs.append(temp)

# ------------------------------------------------------------
# 5. COMBINE ALL FILES
# ------------------------------------------------------------
df = pd.concat(dfs, ignore_index=True)

# Free memory
del dfs
gc.collect()

print("\n========================================")
print("RAW DATASET")
print("========================================")
print("Shape:", df.shape)

print("\nClass distribution:")
print(df["attack_cat"].value_counts())

print("\nNumber of classes:", df["attack_cat"].nunique())

# ------------------------------------------------------------
# 6. REMOVE UNNECESSARY COLUMNS
# ------------------------------------------------------------
# IP addresses and timestamps are removed because:
# - They can cause high-cardinality problems
# - They are not useful as direct ML features
# Label is removed because attack_cat is our target

drop_cols = [
    "srcip",
    "dstip",
    "Stime",
    "Ltime",
    "Label"
]

df = df.drop(columns=drop_cols)

print("\nColumns after removal:", df.shape[1])

# ------------------------------------------------------------
# 7. TRAIN / TEST SPLIT
# ------------------------------------------------------------
# IMPORTANT:
# Split BEFORE balancing to prevent data leakage.
# Test data remains completely untouched.

print("\nCreating train/test split...")

train_df, test_df = train_test_split(
    df,
    test_size=0.20,
    random_state=42,
    stratify=df["attack_cat"]
)

# Free original dataframe
del df
gc.collect()

print("\n========================================")
print("BEFORE BALANCING")
print("========================================")

print("Training rows:", len(train_df))
print("Testing rows:", len(test_df))

print("\nTraining class distribution:")
print(train_df["attack_cat"].value_counts())

# ------------------------------------------------------------
# 8. TARGET COUNTS FOR BALANCING
# ------------------------------------------------------------
# Final training dataset = 660,000 rows
#
# Normal           100,000
# Generic          100,000
# Exploits         100,000
# Fuzzers          100,000
# DoS              100,000
# Reconnaissance   100,000
# Analysis          20,000
# Backdoors         20,000
# Shellcode         10,000
# Worms             10,000
#
# This uses:
# - Undersampling for large classes
# - Oversampling for small classes
#
# NO SMOTE / SMOTENC
# Therefore much lower memory usage.

targets = {
    "Normal": 100000,
    "Generic": 100000,
    "Exploits": 100000,
    "Fuzzers": 100000,
    "DoS": 100000,
    "Reconnaissance": 100000,
    "Analysis": 20000,
    "Backdoors": 20000,
    "Shellcode": 10000,
    "Worms": 10000
}

# ------------------------------------------------------------
# 9. BALANCE TRAINING DATA
# ------------------------------------------------------------
print("\n========================================")
print("BALANCING TRAINING DATA")
print("========================================")

balanced_parts = []

for cls, target_count in targets.items():

    class_data = train_df[train_df["attack_cat"] == cls]

    actual_count = len(class_data)

    if actual_count == 0:
        raise ValueError(
            f"ERROR: Class '{cls}' was not found in the dataset."
        )

    # Undersampling if class has more records
    # Oversampling if class has fewer records
    replace = actual_count < target_count

    sampled_data = class_data.sample(
        n=target_count,
        replace=replace,
        random_state=42
    )

    balanced_parts.append(sampled_data)

    method = "Oversampling" if replace else "Undersampling"

    print(
        f"{cls:18s} : "
        f"{actual_count:8,d} -> "
        f"{target_count:8,d}  ({method})"
    )

# ------------------------------------------------------------
# 10. COMBINE BALANCED CLASSES
# ------------------------------------------------------------
balanced_train = pd.concat(
    balanced_parts,
    ignore_index=True
)

# Free unnecessary objects
del balanced_parts
del train_df
gc.collect()

# Shuffle balanced dataset
balanced_train = balanced_train.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

# ------------------------------------------------------------
# 11. FINAL DISTRIBUTIONS
# ------------------------------------------------------------
print("\n========================================")
print("FINAL BALANCED TRAINING DATA")
print("========================================")

print("Balanced shape:", balanced_train.shape)

print("\nBalanced class distribution:")
print(
    balanced_train["attack_cat"]
    .value_counts()
)

print("\n========================================")
print("TEST DATA - UNTOUCHED")
print("========================================")

print("Test shape:", test_df.shape)

print("\nTest class distribution:")
print(
    test_df["attack_cat"]
    .value_counts()
)

# ------------------------------------------------------------
# 12. SAVE BALANCED TRAIN + ORIGINAL TEST
# ------------------------------------------------------------
balanced_path = os.path.join(
    dataset_path,
    "UNSW_NB15_balanced_train.csv"
)

test_path = os.path.join(
    dataset_path,
    "UNSW_NB15_test.csv"
)

print("\nSaving balanced training dataset...")

balanced_train.to_csv(
    balanced_path,
    index=False
)

print("Balanced training dataset saved.")

print("\nSaving test dataset...")

test_df.to_csv(
    test_path,
    index=False
)

print("Test dataset saved.")

# ------------------------------------------------------------
# 13. FINAL SUMMARY
# ------------------------------------------------------------
print("\n\n================================================")
print("             PIPELINE COMPLETED")
print("================================================")

print("\nBalanced training dataset:")
print(balanced_path)

print("\nTest dataset:")
print(test_path)

print("\nTraining rows:", len(balanced_train))
print("Testing rows :", len(test_df))

print("\nNumber of classes:",
      balanced_train["attack_cat"].nunique())

print("\nFinal training distribution:")
print(
    balanced_train["attack_cat"]
    .value_counts()
    .sort_index()
)

print("\n================================================")
print("DONE")
print("================================================")
```

```python
# ============================================================
# STEP 2: LOAD BALANCED DATA + INSPECT FEATURES
# ============================================================

import pandas as pd
import numpy as np
import os
import gc

dataset_path = "/content/drive/MyDrive/UNSW-NB15"

balanced_path = os.path.join(
    dataset_path,
    "UNSW_NB15_balanced_train.csv"
)

test_path = os.path.join(
    dataset_path,
    "UNSW_NB15_test.csv"
)

# ------------------------------------------------------------
# LOAD DATA
# ------------------------------------------------------------

print("Loading balanced training data...")

train_df = pd.read_csv(
    balanced_path,
    encoding="latin1"
)

print("Training data loaded.")

print("\nLoading test data...")

test_df = pd.read_csv(
    test_path,
    encoding="latin1"
)

print("Test data loaded.")

# ------------------------------------------------------------
# BASIC INFORMATION
# ------------------------------------------------------------

print("\n========================================")
print("DATASET SHAPES")
print("========================================")

print("Training shape:", train_df.shape)
print("Test shape    :", test_df.shape)

print("\n========================================")
print("COLUMN INFORMATION")
print("========================================")

print("Number of columns:", len(train_df.columns))

print("\nColumns:")
for i, col in enumerate(train_df.columns, start=1):
    print(f"{i:2d}. {col}")

# ------------------------------------------------------------
# DATA TYPES
# ------------------------------------------------------------

print("\n========================================")
print("DATA TYPES")
print("========================================")

print(train_df.dtypes)

# ------------------------------------------------------------
# CATEGORICAL COLUMNS
# ------------------------------------------------------------

categorical_cols = train_df.select_dtypes(
    include=["object"]
).columns.tolist()

print("\n========================================")
print("CATEGORICAL COLUMNS")
print("========================================")

print(categorical_cols)

# ------------------------------------------------------------
# NUMERICAL COLUMNS
# ------------------------------------------------------------

numerical_cols = [
    col for col in train_df.columns
    if col not in categorical_cols
]

print("\n========================================")
print("NUMERICAL COLUMNS")
print("========================================")

print("Number of numerical columns:",
      len(numerical_cols))

print(numerical_cols)

# ------------------------------------------------------------
# MISSING VALUES
# ------------------------------------------------------------

print("\n========================================")
print("MISSING VALUES")
print("========================================")

missing = train_df.isnull().sum()

missing = missing[missing > 0].sort_values(
    ascending=False
)

if len(missing) == 0:
    print("No missing values found.")
else:
    print(missing)

# ------------------------------------------------------------
# TARGET CHECK
# ------------------------------------------------------------

print("\n========================================")
print("TARGET")
print("========================================")

print("Target column: attack_cat")

print("\nTraining target distribution:")
print(
    train_df["attack_cat"]
    .value_counts()
    .sort_index()
)

print("\nTest target distribution:")
print(
    test_df["attack_cat"]
    .value_counts()
    .sort_index()
)

print("\n========================================")
print("STEP 2 COMPLETED")
print("========================================")
```

```python
# ============================================================
# STEP 3: CLEAN + ENCODE UNSW-NB15
# ============================================================

import pandas as pd
import numpy as np
import gc

from sklearn.preprocessing import OrdinalEncoder

print("Starting preprocessing...\n")

# ------------------------------------------------------------
# 1. SEPARATE FEATURES AND TARGET
# ------------------------------------------------------------

X_train = train_df.drop(columns=["attack_cat"]).copy()
y_train = train_df["attack_cat"].copy()

X_test = test_df.drop(columns=["attack_cat"]).copy()
y_test = test_df["attack_cat"].copy()

print("Training features:", X_train.shape)
print("Test features    :", X_test.shape)

# ------------------------------------------------------------
# 2. IDENTIFY CATEGORICAL COLUMNS
# ------------------------------------------------------------

categorical_cols = [
    "dsport",
    "proto",
    "state",
    "service",
    "ct_ftp_cmd"
]

# Everything else is numerical
numerical_cols = [
    col for col in X_train.columns
    if col not in categorical_cols
]

print("\nCategorical columns:")
print(categorical_cols)

print("\nNumerical columns:", len(numerical_cols))

# ------------------------------------------------------------
# 3. CLEAN CATEGORICAL COLUMNS
# ------------------------------------------------------------

print("\nCleaning categorical columns...")

for col in categorical_cols:

    X_train[col] = (
        X_train[col]
        .fillna("missing")
        .astype(str)
        .str.strip()
    )

    X_test[col] = (
        X_test[col]
        .fillna("missing")
        .astype(str)
        .str.strip()
    )

# ------------------------------------------------------------
# 4. CONVERT NUMERICAL COLUMNS TO NUMERIC
# ------------------------------------------------------------

print("Converting numerical columns...")

for col in numerical_cols:

    X_train[col] = pd.to_numeric(
        X_train[col],
        errors="coerce"
    )

    X_test[col] = pd.to_numeric(
        X_test[col],
        errors="coerce"
    )

# ------------------------------------------------------------
# 5. HANDLE INFINITE VALUES
# ------------------------------------------------------------

print("Handling infinite values...")

X_train[numerical_cols] = (
    X_train[numerical_cols]
    .replace([np.inf, -np.inf], np.nan)
)

X_test[numerical_cols] = (
    X_test[numerical_cols]
    .replace([np.inf, -np.inf], np.nan)
)

# ------------------------------------------------------------
# 6. FILL NUMERICAL MISSING VALUES
#    MEDIANS ARE CALCULATED ONLY FROM TRAINING DATA
# ------------------------------------------------------------

print("Filling numerical missing values...")

train_medians = X_train[numerical_cols].median()

X_train[numerical_cols] = (
    X_train[numerical_cols]
    .fillna(train_medians)
)

X_test[numerical_cols] = (
    X_test[numerical_cols]
    .fillna(train_medians)
)

# ------------------------------------------------------------
# 7. ORDINAL ENCODING
# ------------------------------------------------------------

print("Encoding categorical features...")

encoder = OrdinalEncoder(
    handle_unknown="use_encoded_value",
    unknown_value=-1
)

X_train[categorical_cols] = encoder.fit_transform(
    X_train[categorical_cols]
)

X_test[categorical_cols] = encoder.transform(
    X_test[categorical_cols]
)

# ------------------------------------------------------------
# 8. CONVERT TO FLOAT32
#    REDUCES MEMORY USAGE
# ------------------------------------------------------------

X_train = X_train.astype(np.float32)
X_test = X_test.astype(np.float32)

# ------------------------------------------------------------
# 9. RESET INDICES
# ------------------------------------------------------------

X_train.reset_index(drop=True, inplace=True)
X_test.reset_index(drop=True, inplace=True)

y_train.reset_index(drop=True, inplace=True)
y_test.reset_index(drop=True, inplace=True)

# ------------------------------------------------------------
# 10. CHECK FOR REMAINING NaN / INF
# ------------------------------------------------------------

print("\n========================================")
print("FINAL DATA CHECK")
print("========================================")

print("Training shape:", X_train.shape)
print("Test shape    :", X_test.shape)

print("\nTraining NaN values:",
      X_train.isna().sum().sum())

print("Test NaN values:",
      X_test.isna().sum().sum())

print("\nTraining infinite values:",
      np.isinf(X_train.to_numpy()).sum())

print("Test infinite values:",
      np.isinf(X_test.to_numpy()).sum())

# ------------------------------------------------------------
# 11. TARGET ENCODING
# ------------------------------------------------------------

class_names = sorted(y_train.unique())

class_to_id = {
    cls: i
    for i, cls in enumerate(class_names)
}

id_to_class = {
    i: cls
    for cls, i in class_to_id.items()
}

y_train_encoded = y_train.map(class_to_id).astype(np.int64)
y_test_encoded = y_test.map(class_to_id).astype(np.int64)

print("\n========================================")
print("CLASS MAPPING")
print("========================================")

for class_id, class_name in id_to_class.items():
    print(f"{class_id}: {class_name}")

# ------------------------------------------------------------
# 12. FINAL SHAPES
# ------------------------------------------------------------

print("\n========================================")
print("PREPROCESSING COMPLETED")
print("========================================")

print("X_train:", X_train.shape)
print("y_train:", y_train_encoded.shape)

print("X_test :", X_test.shape)
print("y_test :", y_test_encoded.shape)

print("\nTraining class IDs:")
print(y_train_encoded.value_counts().sort_index())

print("\nTest class IDs:")
print(y_test_encoded.value_counts().sort_index())

# ------------------------------------------------------------
# 13. FREE ORIGINAL DATAFRAMES
# ------------------------------------------------------------

del train_df
del test_df
del y_train
del y_test
gc.collect()

print("\nOriginal dataframes cleared from memory.")
print("Ready for the next step.")
```
